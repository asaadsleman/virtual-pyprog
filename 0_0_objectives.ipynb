# مقدمة:

مرحباً بالجميع!

يسعدني أن أقدّم لكم دورة سترافقكم في خطواتكم الأولى في عالم علوم الحاسوب الرائع وأساسيات البرمجة.

### لماذا نتعلّم البرمجة؟

في عالم الهندسة والعلوم اليوم، ليست البرمجة وعلوم الحاسوب مجرد أدوات تقنية، بل هي **طريقة تفكير**.

سواءً كنت مهندساً يرغب في تحليل البيانات، أو فيزيائياً يبني محاكاة، أو عالم أحياء يدرس العمليات المعقدة، فإن الجميع مطالبون بفهم كيفية صياغة المشكلات رياضيًّا وتصميم حلول فعّالة لها.

ستساعدك هذه الدورة على تطوير هذه القدرة: التفكير المنهجي والدقيق والإبداعي، وترجمة أفكارك إلى برمجيات فعّالة. **حوّل الفكرة إلى واقع**. بايثون، اللغة التي نتعلمها في هذه الدورة، متوفرة في كل مكان: من قيادة الروبوتات والطابعات ثلاثية الأبعاد، مرورًا بتوقعات الطقس والبحث الجيني، ووصولًا إلى الذكاء الاصطناعي وتوليد الصوّر.

### أهداف ومحتويات الدورة
تهدف هذه الدورة إلى هدفين:
1. **مقدمة إلى لغة برمجة بايثون** - لغة حديثة وبسيطة وقوية سترافقنا طوال الدورة وستكون أساسًا متينًا لرحلتك المستمرة في عالم البرمجة. بالإضافة إلى ذلك، **تطوير التفكير الخوارزمي ومهارات البرمجة العامة** - كيفية تحليل المشكلة إلى مكوناتها الأساسيّة، والتخطيط لحل منهجي، وتنفيذه بفعالية.

٢. **التعرف على المفاهيم والأفكار الأساسية في علوم الحاسوب** - إن الإلمام بـ"اللغة والثقافة العلمية" لهذا المجال سيُسهّل عليك مستقبلًا توسيع معرفتك بعلوم الحاسوب، من خلال التعرّف على مواضيع غير مُغطاة في هذه الدورة.

تُركّز الدورة على الجانب **المنطقي والخوارزمي** للبرمجة وعلوم الحاسوب، ولا تتناول تشغيل الأجهزة نفسها.

يُقسّم المحتوى إلى طبقتين:
- **الطبقة الأولى** تُقدّم أساسيات البرمجة بلغة بايثون والمفاهيم الأساسية في علوم الحاسوب.
- **الطبقة الثانية** تُعالج مواضيع أكثر تقدمًا - التكرار، والبرمجة كائنية التوجه (OOP)، ومكتبات معالجة الصور وتحليل البيانات.

### نبذة عن الذكاء الاصطناعي وكتابة الأكواد البرمجية
لا شك أن نماذج الذكاء الاصطناعي قادرة الآن على حل أي تمرين تقريبًا في الدورة.
- لكن لا تغرِ باستخدامها فورًا. فخبرتك الحقيقية في التعامل مع المشكلة، والتفكير فيها، وارتكاب الأخطاء، وحلها - هي ما يبني فهمك.
- فقط بعد مواجهة السؤال بصدق، يمكنك استخدام أدوات خارجية - بما في ذلك الذكاء الاصطناعي - للتحقق والفهم والتعلم.
